```python
import pandas as pd

def add_edge(mat, i, j):
    mat[i][j] = 1
    mat[j][i] = 1
lines = {
        'Central': ['Notting Hill Gate', 'Queensway', 'Lancaster Gate', 'Marble Arch', 'Bond Street'],
        'Jubilee': ['Baker Street', 'Bond Street', 'Green Park','Westminster', 'Waterloo'],
        'Victoria': ['Green Park', 'Oxford Circus', 'Warren Street', 'Euston', 'King’s Cross St Pancras'],
        'Northern': ['Leicester Square', 'Tottenham Court Road', 'Goodge Street', 'Warren Street', 'Euston'],
        'Piccadilly': ['Green Park', 'Piccadilly Circus', 'Leicester Square', 'Covent Garden', 'Holborn']
    }

all_stations = sorted({station for stations in lines.values() for station in stations})
V = len(all_stations)

station_to_index = {station: i for i, station in enumerate(all_stations)}
mat = [[0] * V for _ in range(V)]

for stations in lines.values():
  for i in range(len(stations) - 1):
      add_edge(mat,
        station_to_index[stations[i]],
        station_to_index[stations[i+1]])

  # Convert matrix into DataFrame
df = pd.DataFrame(mat, index=all_stations, columns=all_stations)
print("Adjacency Matrix:\n")
df
```

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

lines = {
    'Central': ['Notting Hill Gate', 'Queensway', 'Lancaster Gate', 'Marble Arch', 'Bond Street'],
    'Jubilee': ['Baker Street', 'Bond Street', 'Green Park','Westminster', 'Waterloo'],
    'Victoria': ['Green Park', 'Oxford Circus', 'Warren Street', 'Euston', 'King’s Cross St Pancras'],
    'Northern': ['Leicester Square', 'Tottenham Court Road', 'Goodge Street', 'Warren Street', 'Euston'],
    'Piccadilly': ['Green Park', 'Piccadilly Circus', 'Leicester Square', 'Covent Garden', 'Holborn']
}

# Build graph
for line, stations in lines.items():
    G.add_nodes_from(stations)
    G.add_edges_from([(stations[i], stations[i+1]) for i in range(len(stations) - 1)])

print("The graph is connected:", nx.is_connected(G))

# identifying interchange stations
station_count = {} #dictionary to track stations count traversed by a line
for line, stations in lines.items():
    for s in stations:
        station_count[s] = station_count.get(s, 0) + 1
interchanges = [s for s, c in station_count.items() if c > 1] # logic to identify interchange stations if stations appear in more than 1 tube line


#layout
pos = nx.spring_layout(G)
plt.figure(figsize=(12,8))

# track repeated edges for concurrent lines from one station
edge_seen = {} #dictionary to track occurrence of concurrent lines
for line, stations in lines.items():
    line_index = list(lines.keys()).index(line)

    # coloring the edges to represent tube lines
    for i in range(len(stations)-1):
        key = tuple(sorted((stations[i], stations[i+1])))
        #logic to consider and represent concurrent tube lines
        edge_seen[key] = edge_seen.get(key, 0) + 1
        style = "solid" if edge_seen[key] == 1 else "dashed" #solid if there exist only a single line between stations, dashed if multiple lines

        nx.draw_networkx_edges(
            G, pos,
            edgelist=[(stations[i], stations[i+1])],
            width=2,
            alpha=0.5,
            edge_color=[plt.cm.tab10(line_index)],
            style=style
        )

    # original node drawing
    nx.draw_networkx_nodes(
        G, pos, nodelist=stations,
        node_color=plt.cm.tab10(line_index),
        node_size=300,
        label=line
    )

    # original labels
    nx.draw_networkx_labels(
        G, pos,
        labels={station: station for station in stations},
        font_size=9,
        font_color='black'
    )

# white halo circles for interchange stations
nx.draw_networkx_nodes(
    G, pos,
    nodelist=interchanges,
    node_color="white",
    node_size=500,
    edgecolors="black",
    linewidths=1.5
)

plt.title('London Tube Network Example')
plt.legend(scatterpoints=1)
plt.axis('off')
plt.show()
```
