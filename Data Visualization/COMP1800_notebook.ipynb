# **Automatic Segmentation on the basis of specified thresholds.**

```python
#segment limits based on total visitors
segment_limits = {
    0: 'Low Volume',     # lower limits for each segment in ascending order
    50_000: 'Medium Volume',
    100_000: 'High Volume'
}
segments = {}
limits = list(segment_limits.keys())
for i in range(len(limits)):          #code to decide which column are included in the segment
    if i < len(limits) - 1:
        included = (limits[i] <= data.sum()) & (data.sum() < limits[i + 1])
    else:
        included = (limits[i] <= data.sum())
    segment_name = segment_limits[limits[i]]
    segment_columns = data.columns[included]
    segments[segment_name] = list(segment_columns)

#code to initialize the venues into their volume segments
for segment_name in reversed(segments):
    print(f'{segment_name.lower().replace(" ", "_")}_volume = {segments[segment_name]}')

for segment_name in reversed(segments):
    selected = segments[segment_name]
    plt.figure(figsize=(10, 10))
    x_pos = np.arange(len(data[selected].columns))
    plt.bar(x_pos, data[selected].sum(), align='center')
    plt.xticks(x_pos, data[selected].columns)
    plt.xlabel('Venues', fontsize=18)
    plt.ylabel('Number of Visitors', fontsize=18)
    plt.title(segment_name + ' Total Visitors in each Venue', fontsize=20)
    plt.show()
```

# **Improving the plot by using colors to identify the segments. Green = High Volume, Yellow = Medium Volume, Red = Low Volume**

```python
# segment names must match those in segment_limits
segment_colours = {
    'Low Volume': 'red',
    'Medium Volume': 'yellow',
    'High Volume': 'green'
}

# Assign colours to each venue
colours = []
for store in data.sum().index:
    for segment_name in segments:
        if store in segments[segment_name]:
            colours.append(segment_colours[segment_name])
            break

# Plotting the final figure with minimum eye travel principle and other best design prinicples
plt.figure(figsize=(15, 10))
x_pos = np.arange(len(data.columns))
plt.bar(x_pos, data.sum()[data.columns], color=colours, align='center')
plt.gca().set_prop_cycle(None)
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.xticks(x_pos, data.columns)
plt.xlabel('Venues', fontsize=12,loc='right')
plt.ylabel('Total Visitors', fontsize=12, loc='top', rotation=0, labelpad=-30)
plt.title('Total Visitors in each Venue',fontsize=16,loc='left', pad=20)
plt.plot([0, 35], [100_000, 100_000], linestyle='--', linewidth=2, color='black')
plt.plot([0, 35], [50_000, 50_000], linestyle='--', linewidth=2, color='black')
plt.tight_layout()
plt.show()
```

# **Interactive design to explore high volume venues**

**The interactive plot for high volume venues provides the functionality to hover over the data points to see different details of data points across the timeline and switch if off if not desired, it also has zoom tool to scrutinize interested venue and time frame within the plot. It also has reset, undo and redo tools to help user fiddle around with the plots and undo the changes. USer can also zoom in and out using the mouse wheel. Most importantly, it also allows multiple legends to dim out on the basis of what labels to focus. Since there are multple venues, to only focus at user's desired venue, just clicking on other venues dims them.**

```python
import pandas as pd
import hvplot.pandas
import holoviews as hv
hv.extension('bokeh')

selected = high_volume
period = 28
rolling_average = data.rolling(window=period).mean()

# Step 1: Create the base plot for the raw, noisy daily data
raw_plot = data[selected].hvplot.line(
    frame_height=600, frame_width=700,
    line_width=0.5, alpha=0.8,
    xlabel='Date', ylabel='Daily Visitors',
    title='Anomaly Venues (with 28-Day Rolling Average)',
    ylim=(0, None), # ylim=(0, None) to replicate plt.ylim(ymin=0)
    tools=['undo', 'redo'] # Adding interactive tools from the lecture
)

# Step 2: Create the overlay plot for the rolling average
# Using a thicker line to emphasise the trend
rolling_plot = rolling_average[selected].hvplot.line(
    line_width=2
)

# Step 3: Compose the plots together
# The '*' operator overlays multiple plots on top of each other
interactive_plot = raw_plot * rolling_plot

# Plotting
interactive_plot
```

Although the main focus is on high and medium volume venues, the anomalies in the venues need to be identified and looked at more closely. The possible reason for anomalies is shown by above plot. They are the same ones identified through the distribution plot.

```python
anomalous_venues = ['LLJ', 'KLZ', 'MXL', 'XDV', 'KFO', 'LXE', 'NRB', 'LRT']
plt.figure(figsize=(12, 7))

# 1. Reducing Excessive Colour
# Plotting all standard venues in the background using light grey and thin lines
for column in data.columns:
    if column not in anomalous_venues:
        plt.plot(data.index, data[column], color='lightgray', linewidth=0.5)

# Getting the default matplotlib colour cycle to colour the anomalies
default_colours = plt.rcParams['axes.prop_cycle'].by_key()['color']

# 2. Highlighting the Anomalies and to Minimize Eye-Travel
for i, name in enumerate(anomalous_venues):
    colour = default_colours[i % len(default_colours)]

    # Plot the anomalous venues with thicker lines and bright colours
    plt.plot(data.index, data[name], color=colour, linewidth=2)

    # To avoid a disconnected legend, put text labels directly next to the lines
    last_valid_date = data[name].last_valid_index()
    last_value = data[name].loc[last_valid_date]

    plt.text(x=last_valid_date, y=last_value, s=f' {name}',
             fontsize=11, color=colour, fontweight='bold')

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 4. Format titles and labels to minimize eye-travel
plt.xlabel('Date', loc='right')
plt.ylabel('Daily Visitors', loc='top', rotation=0, labelpad=-30)
plt.title('Daily Visitors per Venue (Highlighting Anomalies)', fontsize=16, loc='left', pad=20)
plt.ylim(ymin=0)
plt.tight_layout()
plt.show()
```

# **Exploring Seasonality/Autocorrelation**

```python
data= pd.read_csv('https://raw.githubusercontent.com/ChrisWalshaw/DataViz/master/Data/001483674/VenueDailyVisitors.csv', index_col = 0)
data.index = pd.to_datetime(data.index)
selected = data.columns[data.sum() > 50_000]
row = 0
col = 0
fig, axes = plt.subplots(figsize=(10, 10), nrows=7, ncols=5)
fig.suptitle('Autocorrelation plots for all venues', fontsize=20, position=(0.5, 1.0))
for name in data.columns:
    sub = pd.plotting.autocorrelation_plot(data[name], axes[row, col])
    sub.set_title('Venue ' + name, fontsize=10)
    sub.xaxis.label.set_visible(False)
    sub.yaxis.label.set_visible(False)
    col += 1
    if col == 5:
        row += 1
        col = 0
plt.subplots_adjust(wspace=0.8, hspace=0.8)
plt.show()
```

Too many subplots in the plot causing information overload. So, lets focus only on high and medium volume venues as the company prioritizes.

```python
total_visitors = data.sum()
high_med_venues = total_visitors[total_visitors >= 50000].index.tolist() #to include only the high and medium volume venues

row = 0
col = 0

fig, axes = plt.subplots(figsize=(12, 9), nrows=3, ncols=4)
fig.suptitle('Autocorrelation of High & Medium Volume Venues (Seasonality)', fontsize=20, position=(0.5, 1.0))

for name in high_med_venues:
    sub = pd.plotting.autocorrelation_plot(data[name], axes[row, col])
    sub.set_title('Product ' + name, fontsize=10)
    sub.xaxis.label.set_visible(False)
    sub.yaxis.label.set_visible(False)

    col += 1
    if col == 4:
        row += 1
        col = 0

plt.subplots_adjust(wspace=0.4, hspace=0.6)
plt.show()
```

# **Interactive Bubble Plot of Venue Demographics(Venue Spend vs Venue Age with bubble size as Total Vistors volume)**

**The interactive bubble plot for Venue demographics lets the reader at glance look at the high level correlation between two attributes but at the same time scaling the data points adds another layer to the information conveyed about the volume component for each venues. The reader has the necessary tools to explore and investigate like box zoom, mouse wheel zoom, hover tool, undo and redo for convenience.**

```python
import pandas as pd
import hvplot.pandas

#Setting bubble size using the scaling factor according to volume of daily visitors
summary_data['BubbleSize'] = summary_data['DailyVisitors'] * 0.01

#To make interactive bubble plot using hvplot
bubble_plot = summary_data.hvplot.scatter(
    x='VenueAge',
    y='VenueSpend',
    size='BubbleSize',
    title='Venue Age vs Venue Spend (Bubble Size = Total Visitors Daily)',
    xlabel='Average Venue Age',
    ylabel='Average Venue Spend (£)',
    alpha=0.5,     #for bubble transparency
    padding=0.1,
    hover_cols='all',      #hovertool displays all variables
    frame_height=500,
    frame_width=600,
    tools=['undo', 'redo']
)

# Displaying the plot
bubble_plot
```
